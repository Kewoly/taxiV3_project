# Entraînement DQN — Taxi 15×10

Ce notebook lance l'entraînement du modèle DQN sur l'environnement Taxi 15×10 multi-passagers.

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

from src.envs import LargeTaxiEnv
from src.agent import DQNTaxiAgent
import numpy as np
```

## 1. Initialiser l'environnement

```python
env = LargeTaxiEnv(render_mode=None, grid_width=15, grid_height=10, n_passengers=2)
print(f"Environnement : {env.__class__.__name__}")
print(f"Grille : {env.grid_width}×{env.grid_height}")
print(f"Passagers : {env.n_passengers}")
print(f"Observation space : {env.observation_space}")
print(f"Action space : {env.action_space}")
```

## 2. Créer l'agent DQN

```python
agent = DQNTaxiAgent(
    env=env,
    learning_rate=3e-4,
    buffer_size=100_000,
    batch_size=128,
    gamma=0.99,
    target_update_interval=1000,
    exploration_final_eps=0.02,
    exploration_fraction=0.1,
    tensorboard_log="../logs",
)
print("Agent DQN créé")
```

## 3. Lancer l'entraînement

⚠️ Cela peut prendre plusieurs minutes selon le nombre de timesteps.

```python
print("Lancement de l'entraînement...")
print("TensorBoard : tensorboard --logdir ../logs")

agent.train(
    total_timesteps=300_000,
    checkpoint_dir="../training/checkpoints",
)

print("Entraînement terminé !")
```

## 4. Sauvegarder le modèle final

```python
agent.save("../training/checkpoints/dqn_taxi_final.zip")
env.close()
print("Modèle sauvegardé !")
```
